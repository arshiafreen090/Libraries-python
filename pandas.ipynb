## Some Important Series Methods

```python
# astype
# between
# clip
# drop_duplicates
# isnull
# dropna
# fillna
# isin
# apply
# copy
```

### `astype()`
This method is used to cast a Series to a specified `dtype` (data type). It's useful for converting data types for operations or memory optimization.

```python
# Example for astype()
# Convert 'marks_series' to float type
float_marks = marks_series.astype(float)
print('Original Series dtype:', marks_series.dtype)
print('Converted Series dtype:', float_marks.dtype)
display(float_marks)
```

### `between()`
This method returns a boolean Series indicating whether each element in the Series is between two specified values (inclusive by default). It's great for filtering data within a range.

```python
# Example for between()
# Check marks between 95 and 98
marks_between = marks_series.between(95, 98)
print('Marks Series values between 95 and 98:')
display(marks_series[marks_between])
```

### `clip()`
This method is used to trim values at input thresholds. Values above the upper bound will be set to the upper bound, and values below the lower bound will be set to the lower bound.

```python
# Example for clip()
# Clip marks_series values between 98 and 99
clipped_marks = marks_series.clip(lower=98, upper=99)
print('Original Marks Series:')
display(marks_series)
print('\nClipped Marks Series (lower=98, upper=99):')
display(clipped_marks)
```

### `drop_duplicates()`
This method removes duplicate values from a Series, returning a Series with unique elements. By default, it keeps the first occurrence of each unique value.

```python
# Example for drop_duplicates()
dup_series = pd.Series([1, 2, 2, 3, 4, 3, 5])
print('Original Series with duplicates:')
display(dup_series)
print('\nSeries after dropping duplicates:')
display(dup_series.drop_duplicates())
```

### `isnull()`
This method returns a boolean Series indicating whether each element is `NaN` (Not a Number) or `None`. It's essential for identifying missing data.

```python
# Example for isnull()
import numpy as np
missing_series = pd.Series([1, np.nan, 3, None, 5])
print('Series with missing values:')
display(missing_series)
print('\nBoolean Series indicating null values:')
display(missing_series.isnull())
```

### `dropna()`
This method removes missing values (NaN/None) from the Series. It returns a new Series with the non-missing data.

```python
# Example for dropna()
import numpy as np
missing_series = pd.Series([1, np.nan, 3, None, 5])
print('Original Series with missing values:')
display(missing_series)
print('\nSeries after dropping null values:')
display(missing_series.dropna())
```

### `fillna()`
This method fills missing values (NaN/None) in the Series with a specified value or method (e.g., forward fill, backward fill).

```python
# Example for fillna()
import numpy as np
missing_series = pd.Series([1, np.nan, 3, None, 5])
print('Original Series with missing values:')
display(missing_series)
print('\nSeries after filling null values with 0:')
display(missing_series.fillna(0))
print('\nSeries after filling null values with forward fill:')
display(missing_series.fillna(method='ffill'))
```

### `isin()`
This method returns a boolean Series indicating whether each element in the Series is contained in a passed sequence of `values`. It's useful for filtering data based on a list of desired items.

```python
# Example for isin()
# Check if marks are 97 or 99
allowed_marks = [97, 99]
is_in_series = marks_series.isin(allowed_marks)
print('Original Marks Series:')
display(marks_series)
print('\nBoolean Series indicating values in [97, 99]:')
display(is_in_series)
print('\nMarks that are either 97 or 99:')
display(marks_series[is_in_series])
```

### `apply()`
This method is used to apply a function to every element in the Series. The function can be a built-in Python function, a lambda function, or a custom function you define.

```python
# Example for apply()
# Apply a lambda function to add 5 to each mark
marks_plus_five = marks_series.apply(lambda x: x + 5)
print('Original Marks Series:')
display(marks_series)
print('\nMarks Series after applying lambda x: x + 5:')
display(marks_plus_five)

# Apply a custom function
def categorize_mark(mark):
    if mark >= 98:
        return 'Excellent'
    elif mark >= 90:
        return 'Good'
    else:
        return 'Average'

categorized_marks = marks_series.apply(categorize_mark)
print('\nMarks Series after applying custom categorization function:')
display(categorized_marks)
```

### `copy()`
This method creates a deep copy of the Series, meaning changes to the new Series will not affect the original Series. This is important to avoid unintended modifications to your data.

```python
# Example for copy()
original_series = pd.Series([10, 20, 30])
copied_series = original_series.copy()

print('Original Series:')
display(original_series)
print('Copied Series:')
display(copied_series)

# Modify the copied series
copied_series[0] = 100

print('\nOriginal Series after modifying copied series:')
display(original_series)
print('Copied Series after modification:')
display(copied_series)
# Notice that the original_series remains unchanged
```
