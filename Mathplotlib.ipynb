## Bar Chart


- Bivariate Analysis

````python
# simple bar chart (default)
children= [10,20,30,40,50]
colors= ['red','pink','blue','black','yellow']
plt.bar(colors, children, color=colors)
````

````python
# color and label
df= pd.read_csv('/content/batsman_season_record.csv')
df
````

````python
plt.bar(np.arange(df.shape[0])-0.2,df["2015"],width=0.2, color='#f777c6')
plt.bar(np.arange(df.shape[0]),df["2016"],width=0.2, color='#7f77f7')
plt.bar(np.arange(df.shape[0])+0.2,df["2017"],width=0.2, color='#abf56e')

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.title('bar graph of batsman')
plt.xlabel('batsman')
plt.ylabel('runs')
plt.show()
````

````python
children= [10,20,30,40,50]
colors= ['#f2464f','#de7ceb','#72b6ed','#0b0c0d','#e1e693']
plt.xticks(np.arange(len(children)),['red red red red ','pink pink pink pink','blue blue blue blue','black black black black','yellow yellow yellow yellow'])
display(plt.bar(colors, children, color=colors))

# to fix this we use
# plt.xticks(rotation='vertical')
# now
plt.xticks(np.arange(len(children)),['red red red red ','pink pink pink pink','blue blue blue blue','black black black black','yellow yellow yellow yellow'], rotation='vertical')
display(plt.bar(colors, children, color=colors))
````

````python
# stacked Bar graph

# Base layer for 2017 (pink)
plt.bar(df['batsman'], df['2017'], color='#f777c6', label='2017')

# Stack 2016 on top of 2017 (purple)
plt.bar(df['batsman'], df['2016'], bottom=df['2017'], color='#7f77f7', label='2016')

# Stack 2015 on top of (2016 + 2017) (light green)
plt.bar(df['batsman'], df['2015'], bottom=df['2016'] + df['2017'], color='#94e3ab', label='2015')
plt.legend()
plt.show()
````

### Colored Scatterplots

````python
iris= pd.read_csv('/content/iris.csv')
iris
iris.sample(5)
````

````python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# coloured scatter plot with cmap and alpha
iris["Species"]= iris["Species"].replace("Iris-setosa","0")
iris["Species"]= iris["Species"].replace("Iris-versicolor","1")
iris["Species"]= iris["Species"].replace("Iris-virginica","2")
iris["Species"] = iris["Species"].astype(int) # Convert to integer type
print(iris)
plt.scatter(iris["SepalLengthCm"],iris["PetalLengthCm"],c=iris["Species"],cmap='jet',alpha=0.7)
plt.colorbar()
plt.xlabel("SepalLengthCm")
plt.ylabel("PetalLengthCm")
plt.show()
````

## Plot size

````python
plt.figure(figsize=(10,5))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
````

## Subplots

- fig= one whole graph where all the differnt graph(ax) inside  
- ax= differnt graphs in one fig

````python
fig,ax= plt.subplots(nrows=2,ncols=1,figsize=(10,10))
batters= pd.read_csv('/content/batter.csv')

ax[0].scatter(batters['avg'],batters['strike_rate'],color='#7f77f7')
ax[1].scatter(batters['avg'],batters['runs'])
ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')


ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
````

````python
fig, ax= plt.subplots(nrows=2,ncols=2,figsize=(10,10))
ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='#7f77f7')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
````

````python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['avg'])
````

````python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

# Define x, y, and create meshgrid for xx, yy, and a sample z
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
xx, yy = np.meshgrid(x, y)
z = np.sin(np.sqrt(xx**2 + yy**2)) # Example: a simple 3D surface

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
plt.show() # Added plt.show() to explicitly display the plot
````
