Plots under this section

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips= sns.load_dataset('tips')
tips.head()
```

 `plt.scatter()` vs `sns.relplot()`

- `sns.relplot()` → it control over the whole fig whereas `plt.scatter()` controls only on axes thats why its legend inside the box.
- `plt.scatter()` → Creates a simple scatter plot.
- `sns.relplot()` → Shows relationships between variables. also it is very custmizatiable.
- `relplot()` supports both **scatter and line plots** using `kind`.
- `relplot()` makes grouping easy using `hue`, `row`, and `col`.


 Axes-Level Functions
- `sns.scatterplot()` → Creates a scatter plot
- `sns.lineplot()` → Creates a line plot

 Figure-Level Functions
- `sns.relplot()` → High-level relationship plot
  - `kind="scatter"` → Scatter plot
  - `kind="line"` → Line plot

**Easy way to remember:**

Axes-level → `scatterplot()`, `lineplot()`

Figure-level → `relplot()`

```python
# scatter plot ---> axes level function
plt.figure(figsize=(5,6))
sns.scatterplot(data=tips,x='total_bill', y='tip', hue='sex',style='time',size='size')
plt.show()
```

```python
# relplot -> figure level -> square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',style='time',size='size')
```

```python
# lineplot
gap=px.data.gapminder()
temp_gap= gap[gap['country']== 'India']
temp_gap
```

```python
# axes level function
sns.lineplot(data=temp_gap, x='year', y='lifeExp')
plt.show()
```

```python
# using relplot
sns.relplot(data=temp_gap,x='year',y='lifeExp', kind='line')
```

```python
# hue --> style
temp_df= gap[gap['country'].isin(['India','Brazil','Japan'])]
temp_df
```

```python
sns.relplot(data=temp_df,x='year',y='lifeExp', kind='line',hue='country',style='country')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot

sns.relplot(data=tips,x='total_bill',y='tip',col='time',hue='sex',kind='scatter')
```

```python
# col wrap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```

### 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram
sns.histplot(data=tips, x='total_bill')
tips
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=2)
```

```python
# hue
sns.displot(data=tips,x='total_bill',kind='hist',hue='sex')
```

```python
# element---> step
sns.displot(data=tips, x='tip', kind='hist',hue='sex',element='step')
```

```python
titanic= sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x="age", hue="sex",kind="hist", element="step")
```

```python
#  faceting using col and row -> not work on histplot function
sns.displot(data=tips, x='tip', kind='hist',col='sex',element='step', color='purple')
```

```python
# kdplot
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate
sns.displot(data=tips, x='total_bill', kind='kde')
plt.title('density vs total bill')
```

```python
# hue---> fill
sns.displot(data=tips, x='total_bill', kind='kde',hue='sex',fill=True)
```

```python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color
sns.displot(data=tips, x='total_bill', y='tip',kind='hist')
# here x and y both are given
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.kdeplot(data=tips, x='total_bill', y='tip')
```

### 3. Matrix Plot

- Heatmap
- Clustermap

```python
# Heatmap

# Plot rectangular data as a color-encoded matrix
plt.figure(figsize=(15,15))
temp_df= gap.pivot(index='country', columns='year',values='lifeExp')
sns.heatmap(temp_df)
```

```python
# annot
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5, cmap='summer')
```

```python
# Clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.

# This function requires scipy to be available.

iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```

```python
# boxplot
```

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

```python
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex')
plt.title('Total Bill Distribution by Day and Sex')
plt.show()
```

```python
# Using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips,y='total_bill')
```

 Violinplot = (Boxplot + KDEplot)

```python
# violinplot
sns.violinplot(data=tips,x='day',y='total_bill')
```
